```python
df = df.drop_duplicates()
print("After Removing Duplicates:", df.shape)
```

```python
categorical_cols = df.select_dtypes(include="object").columns

for col in categorical_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
for col in categorical_cols:
    df[col] = df[col].str.strip()
```

```python
for col in df.columns:
    if "date" in col.lower():
        df[col] = pd.to_datetime(df[col], errors="coerce")
```

```python
df = df.dropna(subset=[col for col in df.columns if "date" in col.lower()])
```

```python
print("\nNull Values After Cleaning:\n")
print(df.isnull().sum())
```

```python
print("\nData Cleaning Completed ✅")
print("Cleaned file saved as: space_missions_cleaned.csv")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("C:/Users/prave/Downloads/space_missions_final.csv")
```

```python
print(df.columns)
```

```python
#  1. Year-wise Launch Count
df['Date'].value_counts().sort_index()
```

```python
# 2. Mission Status Table

status_table = df['MissionStatus'].value_counts()

print(status_table)
```

```python
# 3. Company-wise Launch Count Table
company_table = df['Company'].value_counts().head(10).reset_index()
company_table.columns = ['Company', 'Number_of_Launches']

print(company_table)
```

```python
# 4. Rocket Status Table
rocket_status_table = df['RocketStatus'].value_counts().reset_index()
rocket_status_table.columns = ['RocketStatus', 'Count']

print(rocket_status_table)
```

```python
 # 5. MissionStates Success Failure Table

df.columns = df.columns.str.strip()
pivot_table = df.pivot_table(
    index='Date',
    columns='MissionStatus',
    aggfunc='size',
    fill_value=0
)

print(pivot_table)
```

```python
# 6.Launch Cost Distribution
price_summary = df['Price'].describe()

print(price_summary)
```


<h2 style='font-size:35px; color:White; text-align:center;'>
📊 BIVARIATE ANALYSIS
</h2>


```python
 # 1.Company vs Success Count
successful = df[df['MissionStatus'].str.strip() == 'Success']

company_success = (
    successful
    .groupby('Company')
    .size()
    .reset_index(name='Number_of_Successful_Missions')
    .sort_values(by='Number_of_Successful_Missions', ascending=False)
)

print(company_success.head(10))
```

```python
# 2. Cost vs Status
plt.figure()
sns.boxplot(x='MissionStatus', y='Price', data=df)

```

```python
# 3. Year vs Success Rate
success_rate = df.groupby('Date')['MissionStatus'].apply(lambda x: (x=='Success').mean())

plt.figure()
success_rate.plot()
plt.title("Success Rate Over Years")
plt.ylabel("Success Rate")
plt.show()
```

```python

successful = df[df['MissionStatus'] == 'Success']

company_success = successful['Company'].value_counts()

print(company_success.head(10))

```

```python
# 4.Mission Failed Count in Each Company
failed = df[df['MissionStatus'] == 'Failure']

company_failed = failed['Company'].value_counts()

print(company_failed.head(10))

```


<h2 style='font-size:35px; color:darkorange; text-align:center;'>
Data Visualization
</h2>


```python
df['Date'] = pd.to_datetime(df['Date'])

df['Year'] = df['Date'].dt.year

year_count = df['Year'].value_counts().sort_index()

year_count.plot()
plt.title("Number of Missions Per Year")
plt.xlabel("Year")
plt.ylabel("Mission Count")
plt.show()
```

Insight: Successful missions usually have higher average cost compared to failed missions. This indicates that larger or more complex missions tend to succeed more often.

```python
success_percent = (len(successful) / len(df)) * 100

print("Overall Mission Success Rate:", round(success_percent,2), "%")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
plt.figure()
df["Company"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Companies by Total Missions")
plt.xlabel("Company")
plt.ylabel("Mission Count")
plt.show()
```

Insight: A few major space companies dominate the number of launches, showing their strong presence in the space industry.

```python
plt.figure()
df["MissionStatus"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Mission Status Distribution")
plt.ylabel("")
plt.show()
```

Insight: Most missions are successful, while a smaller percentage of missions end in failure or partial failure.

```python
plt.figure()
df["Company"].value_counts().head(5).plot(kind="pie", autopct="%1.1f%%")
plt.title("Top 5 Companies Share")
plt.ylabel("")
plt.show()
```

Insight: The top 5 companies contribute a large share of the total missions, showing their dominance in the space launch sector.

```python
plt.figure()
df["Random_Performance_score"].plot(kind="hist", bins=20)
plt.title("Performance Score Distribution")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()
```

Insight: Most missions have a moderate performance score, while very high or very low scores are less common.

```python
plt.figure()
df.boxplot(column="Random_Performance_score")
plt.title("Performance Score Boxplot")
plt.show()
```

Insight: The boxplot shows the spread of performance scores and highlights possible outliers among mission performances.

```python
#  Correlation Heatmap
plt.figure()
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

Insight: The heatmap shows relationships between numeric variables such as cost, performance score, and mission counts.


```python
if 'Location' in df.columns:
    plt.figure()
    df['Location'].value_counts().head(10).plot(kind='barh')
    plt.title("Top 10 Launch Locations")
    plt.xlabel("MissionCount")
    plt.show()
```

Insight: The top launch locations handle the majority of space missions. This shows that a few well-developed launch sites are preferred because they have better infrastructure, safety systems, and geographical advantages for launching rockets.

```python
# Cost + Year + Status (Scatter Plot)
plt.figure()
sns.scatterplot(data=df,
                x='Date',
                y='Price',
                hue='MissionStatus')
plt.title("Cost vs Year by Status")
plt.show()
```

Insight: The scatter plot shows that mission costs vary across years, and successful missions appear across different cost ranges.
